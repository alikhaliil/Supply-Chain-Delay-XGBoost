```python
import pandas as pd 
import numpy as np 
df = pd.read_csv('DataCoSupplyChainDataset.csv' , encoding='latin1')
df.head()
```

```python
df.info()
```

```python
pd.set_option('display.max_columns', None)
df.head()
```

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
```

```python
to_drop = [ 'Customer Email', 'Customer Password', 'Customer Fname', 'Customer Lname',
            'Product Description', 'Product Image', 'Customer Street', 'Order Zipcode',
            'Customer Id', 'Order Id', 'Order Item Id', 'Product Card Id', 'Product Category Id']
df.drop(columns=to_drop , inplace=True)
print(df['Late_delivery_risk'].unique())
df.head()
```

```python
financial_cols = [
    'Benefit per order', 'Sales per customer', 'Order Item Discount', 
    'Order Item Discount Rate', 'Order Item Product Price', 
    'Order Item Profit Ratio', 'Sales', 'Order Item Total', 
    'Order Profit Per Order', 'Product Price'
]
df.drop(columns= financial_cols , inplace=True , errors ='ignore')
print(df.columns.tolist())
```

```python
leakage_cols = ['Delivery Status', 'Days for shipping (real)']
df.drop(columns=leakage_cols ,inplace =True ,errors='ignore')
duplicates = df.duplicated().sum()
if duplicates > 0 :
    df.drop_duplicates(inplace=True)
    print(f"done {duplicates}")
print(f"the final shape {df.shape}")
```

```python
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])

df['Order_Month'] = df['order date (DateOrders)'].dt.month
df['Order_Day'] = df['order date (DateOrders)'].dt.dayofweek
df['Order_Hour'] = df['order date (DateOrders)'].dt.hour


cols_to_remove = ['order date (DateOrders)', 'order date (DateOrder)', 'shipping date (DateOrders)']
df.drop(columns=cols_to_remove, inplace=True, errors='ignore')

print("Feature Engineering for dates: SUCCESS")
print(f"Final columns in df: {df.columns.tolist()}")
```

```python
from sklearn.model_selection import train_test_split
x = df.drop(columns=['Late_delivery_risk'])
y = df['Late_delivery_risk']
x_train ,x_test ,y_train,y_test =train_test_split(x,y,test_size=0.2 ,random_state=42,stratify=y)
print(f"Training Features Shape (X_train): {x_train.shape}")
print(f"Training Target Shape (y_train): {y_train.shape}")
print("-" * 40)
print(f"Testing Features Shape (X_test): {x_test.shape}")
print(f"Testing Target Shape (y_test): {y_test.shape}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
region_mean = y_train.groupby(x_train['Order Region']).mean()
country_mean = y_train.groupby(x_train['Order Country']).mean()
state_mean = y_train.groupby(x_train['Order State']).mean()
city_mean = y_train.groupby(x_train['Order City']).mean()
x_train['Geo_Risk_Score'] = x_train['Order City'].map(city_mean)
x_test['Geo_Risk_Score'] = x_test['Order City'].map(city_mean)
x_test['Geo_Risk_Score'] = x_test['Geo_Risk_Score'].fillna(x_test['Order State'].map(state_mean))
x_test['Geo_Risk_Score'] = x_test['Geo_Risk_Score'].fillna(x_test['Order Country'].map(country_mean))
x_test['Geo_Risk_Score'] = x_test['Geo_Risk_Score'].fillna(x_test['Order Region'].map(region_mean))
geo_cols = ['Order Region', 'Order Country', 'Order State', 'Order City', 'Customer City', 'Customer State', 'Customer Country', 'Market']
x_train.drop(columns=[col for col in geo_cols if col in x_train.columns], inplace=True)
x_test.drop(columns=[col for col in geo_cols if col in x_test.columns], inplace=True)
categorical_cols = x_train.select_dtypes(include=['object']).columns.tolist()
for col in categorical_cols:
    le = LabelEncoder()
    x_train[col] = le.fit_transform(x_train[col].astype(str))
    le_dict = dict(zip(le.classes_, le.transform(le.classes_)))
    x_test[col] = x_test[col].astype(str).map(le_dict).fillna(-1).astype(int)
print(f"Number of Text Columns left in x_train: {x_train.select_dtypes(include=['object']).shape[1]}")
print(f"Final x_train shape: {x_train.shape}")
print(f"Final x_test shape: {x_test.shape}")
print("Data is 100% numeric and READY for the Model")
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
xgb_model =XGBClassifier(n_estimators=100,learning_rate=0.1,max_depth=5,random_state=42,eval_metric='logloss')
print("trainig...")
xgb_model.fit(x_train,y_train)
print("training completed")
y_pred = xgb_model.predict(x_test)
accuracy =accuracy_score(y_test,y_pred)
print(f"Initial Model Accuracy: {accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Not Late', 'Late'], 
            yticklabels=['Not Late', 'Late'])
plt.xlabel('Predicted') 
plt.ylabel('Actual')
plt.title('Confusion Matrix - Late Delivery Prediction')
plt.show()
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

importance = xgb_model.get_booster().get_score(importance_type='weight')
keys = list(importance.keys())
values = list(importance.values())

feature_importance_df = pd.DataFrame(data=values, index=keys, columns=["score"]).sort_values(by="score", ascending=True)

plt.figure(figsize=(10, 8))
feature_importance_df.tail(10).plot(kind='barh', color='skyblue', legend=False)
plt.title('Top 10 Important Features - XGBoost')
plt.xlabel('Importance Score (Weight)')
plt.show()

print("Top 5 Driving Features:")
print(feature_importance_df.tail(5).iloc[::-1])
```

```python
x_train.drop(columns=['Order Status'], inplace=True, errors='ignore')
x_test.drop(columns=['Order Status'], inplace=True, errors='ignore')
late_count = (y_train == 1).sum()
not_late_count = (y_train == 0).sum()
ratio = not_late_count / late_count
xgb_model_v2 = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    scale_pos_weight=ratio,
    random_state=42,
    eval_metric='logloss'
)
xgb_model_v2.fit(x_train, y_train)
y_pred_v2 = xgb_model_v2.predict(x_test)
from sklearn.metrics import classification_report, accuracy_score
print(f"New Model Accuracy: {accuracy_score(y_test, y_pred_v2) * 100:.2f}%")
print("\nUpdated Classification Report:")
print(classification_report(y_test, y_pred_v2))
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
import numpy as np
x_train['Shipping_Geo_Interaction'] = x_train['Shipping Mode'] * x_train['Geo_Risk_Score']
x_test['Shipping_Geo_Interaction'] = x_test['Shipping Mode'] * x_test['Geo_Risk_Score']
param_dist = {
    'n_estimators': [100, 200],
    'max_depth': [5, 7, 9],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.7, 0.8, 0.9],
    'colsample_bytree': [0.7, 0.8, 0.9],
    'scale_pos_weight': [ratio] 
}
xgb = XGBClassifier(random_state=42, eval_metric='logloss')
random_search = RandomizedSearchCV(
    xgb, param_distributions=param_dist, 
    n_iter=10, cv=3, scoring='recall', 
    verbose=3, random_state=42, n_jobs=-1
)
random_search.fit(x_train, y_train)
best_model = random_search.best_estimator_
y_pred_tuned = best_model.predict(x_test)

print("-" * 40)
print(f"Best Parameters found: {random_search.best_params_}")
print(f"New Tuned Accuracy: {accuracy_score(y_test, y_pred_tuned) * 100:.2f}%")
print("\nFinal Classification Report (Tuned):")
print(classification_report(y_test, y_pred_tuned))
```
